# Secrets Helper
Use this notebook to add [Databricks Secrets](https://docs.databricks.com/aws/en/security/secrets/) (e.g. for Steam, Reddit API keys) to your workspace so that they may be used by the demo app.

Simply fill in any/all of the secret values below and then use "Run all" to run the entire notebook.
After running the notebook, it is recommended to clear/remove the hardcoded values in the cell below for security; they will still be available via the stored Databricks Secret.

```python
# Fill in your Secret values here.
# If you do not have values for all of them, leave the rest blank.

# Steam
steam_api_key_value = ""

# Reddit
reddit_client_id_value = ""
reddit_client_secret_value = ""
reddit_user_agent_value = ""
```

## Add Secrets to Workspace
You do not need to modify this section.

```python
from databricks.sdk import WorkspaceClient
from databricks.sdk.errors import ResourceAlreadyExists

# Initialize the WorkspaceClient
workspace_client = WorkspaceClient()
secret_scope = "social_listening_app"

# Create the secret scope
try:
    workspace_client.secrets.create_scope(scope=secret_scope)
    print(f"Created secret scope: {secret_scope}")
except ResourceAlreadyExists:
    print(f"Secret scope '{secret_scope}' already exists, continuing...")

# Add secrets to the scope
if steam_api_key_value:
    workspace_client.secrets.put_secret(scope=secret_scope, key="steam_api_key", string_value=steam_api_key_value)
    print("Added steam_api_key")

if reddit_client_id_value:
    workspace_client.secrets.put_secret(scope=secret_scope, key="reddit_client_id", string_value=reddit_client_id_value)
    print("Added reddit_client_id")

if reddit_client_secret_value:
    workspace_client.secrets.put_secret(scope=secret_scope, key="reddit_client_secret", string_value=reddit_client_secret_value)
    print("Added reddit_client_secret")

if reddit_user_agent_value:
    workspace_client.secrets.put_secret(scope=secret_scope, key="reddit_user_agent", string_value=reddit_user_agent_value)
    print("Added reddit_user_agent")

print("Finished adding secrets.")
```

```python
# Verify Secrets creation
from databricks.sdk.errors import ResourceDoesNotExist
try:
    secret_list = workspace_client.secrets.list_secrets(scope=secret_scope)

    print(f"Secrets in scope '{secret_scope}': {len(secret_list)}")
    for secret_metadata in secret_list:
        print(f"  - Key: {secret_metadata.key}")

except ResourceDoesNotExist:
    print(f"Secret scope '{secret_scope}' does not exist.")
```

## Delete Secrets
Use this section to delete/remove the secrets created above, either individually by key name (which you can find from the output of the cell above) or altogether.

```python
secret_key_to_delete = "" # e.g. "steam_api_key". Remember to clear this value afterward if you use this cell.

# Fill in a value for the variable above to delete that secret.
from databricks.sdk import WorkspaceClient
from databricks.sdk.errors import ResourceDoesNotExist

workspace_client = WorkspaceClient()
secret_scope = "social_listening_app"

try:
    workspace_client.secrets.delete_secret(scope=secret_scope, key=secret_key_to_delete)
    print(f"Deleted secret: {secret_scope}/{secret_key_to_delete}")
except ResourceDoesNotExist:
    print(f"Secret key '{secret_key_to_delete}' does not exist; nothing to delete.")
```
